```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder,StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score,roc_auc_score
```

```python
dataset = pd.read_csv("credit_scoring_dataset.csv")
dataset.head(5)
```

```python
q1 = dataset["income"].quantile(0.25)
q3 = dataset["income"].quantile(0.75)
iqr = q3-q1
q1,q3,iqr
```

```python
max_range = q3 + (1.5 * iqr)
min_range = q1 - (1.5 * iqr)
max_range,min_range
```

```python
dataset = dataset[(dataset["income"] >= min_range) & (dataset["income"] <= max_range)]
```

```python
dataset
```

```python
sns.boxplot(dataset)
plt.show()
```

```python
dataset.shape
```

```python
dataset.info()
```

```python
sns.pairplot(dataset,hue="creditworthy")
plt.show()
```

```python
x = dataset.iloc[:,:-1]
y = dataset["creditworthy"]
```

```python
st = StandardScaler()
trans_x = st.fit_transform(x)
x = pd.DataFrame(trans_x,columns=x.columns)
x
```

```python
x_trans = st.fit_transform(x)
```

```python
x = pd.DataFrame(x_trans,columns=x.columns)
x
```

```python
# lo = LogisticRegression()
# score =[] 
# for i in range(1,100):
#     x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=i)
#     lo.fit(x_train,y_train)
#     train_score = lo.score(x_train,y_train)*100
#     test_score = lo.score(x_test,y_test)*100
#     if train_score >= test_score >= train_score-2:
#         score.append([i,"  ",train_score,"  ",test_score])

# score
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=18)
```

```python
lo = LogisticRegression(class_weight={1:3})
```

```python
lo.fit(x_train,y_train)
```

```python
train_score = lo.score(x_train,y_train)*100
test_score = lo.score(x_test,y_test)*100
print("train_score:",train_score,"  ","test_score:",test_score)
```

```python
y_pred = lo.predict(x_test)
for i in y_pred:
    print(i)
```

```python
acc = accuracy_score(y_test,y_pred)*100
acc
```

```python
cm = confusion_matrix(y_test,y_pred)
cm
```

```python
sns.heatmap(cm,annot=True)
plt.show()
```

```python
ps = precision_score(y_test,y_pred)*100
ps
```

```python
rc = recall_score(y_test,y_pred)*100
rc
```

```python
f1 = f1_score(y_test,y_pred)*100
f1
```

```python
y_pred_proba = lo.predict_proba(x_test)[:, 1]
roc_auc = roc_auc_score(y_test, y_pred_proba)*100
roc_auc
```
